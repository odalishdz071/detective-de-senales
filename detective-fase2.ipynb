```python
# 2.2 Diseñar el modelo (CNN 1D)
import tensorflow as tf
from tensorflow.keras import layers, models

# Como la señal es una línea de tiempo usamos capas Conv1D
model = models.Sequential([
    # Capa de entrada: Recibe nuestras señales de 2000 muestras (1 canal)
    layers.Input(shape=(n_samples, 1)),

    # Primera capa convolucional: Busca patrones básicos en la onda
    layers.Conv1D(filters=32, kernel_size=3, activation='relu'),
    layers.MaxPooling1D(pool_size=2), # Reduce el tamaño a la mitad

    # Segunda capa convolucional: Busca patrones más complejos
    layers.Conv1D(filters=64, kernel_size=3, activation='relu'),
    layers.MaxPooling1D(pool_size=2),

    # Aplanamos los datos para pasarlos a las capas densas
    layers.Flatten(),

    # Capa densa (neuronas) para procesar lo aprendido
    layers.Dense(64, activation='relu'),

    # Capa de salida: 4 neuronas (ASK, FSK, PSK, QAM) con Softmax
    layers.Dense(4, activation='softmax')
])

# Mostrar un resumen de la arquitectura
model.summary()
```
